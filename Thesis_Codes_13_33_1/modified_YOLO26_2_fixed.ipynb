```python
from google.colab import drive
drive.mount('/content/drive')

content = '''import torch
import torch.nn as nn

class WeightedConcat(nn.Module):
    def __init__(self, dimension=1, n=2):
        super().__init__()
        self.d = dimension
        self.w = nn.Parameter(torch.ones(n), requires_grad=True)
        self.epsilon = 1e-4

    def forward(self, x):
        w = torch.relu(self.w)
        w = w / (w.sum() + self.epsilon)
        x = [xi * wi for xi, wi in zip(x, w)]
        return torch.cat(x, self.d)

# NOTE: no global monkey-patch here (no `tasks.Concat = WeightedConcat`).
# This class is registered per-run in Colab via:
#   import ultralytics.nn.tasks as tasks
#   tasks.WeightedConcat = WeightedConcat
# which makes it resolvable by NAME for YAML layers explicitly typed
# 'WeightedConcat', while layers typed 'Concat' remain the stock class.
# Safe to reuse unmodified across F1/F2/F3/F4 selective ablation experiments -
# only the YAML determines which single layer becomes weighted.
'''

path = "/content/drive/MyDrive/Thesis/Experiments1/custom_modules_clean.py"
with open(path, "w") as f:
    f.write(content)

print(f"Written to: {path}")

# Verify
with open(path) as f:
    print(f.read())
```

```python
# RECONSTRUCTED CELL — this was missing from the uploaded notebook.
# NEW_YAML_F1 was referenced in the next cell but never defined anywhere,
# which would raise NameError on a fresh run. This mirrors the exact
# pattern used for F2 (Cell 4-5), F3, and F4 below.
#
# F1 = absolute layer 12 = head index 1 (backbone_len=11, so 12-11=1).
# Per the architecture notes, this is the first top-down fusion point
# (P5-upsampled + P4), matching [[-1, 6], 1, 'Concat', [1]] in the stock
# yaml — same source index (6) used for the "P5up + P4" fusion in the
# WeightedFusion(All) architecture. VERIFY this against your actual stock
# yolo26.yaml before training — the assert below will fail loudly if the
# guess is wrong, which is intentional (better a crash here than silently
# modifying the wrong layer).

import yaml, shutil, os

STOCK_YAML = result.stdout.strip().split("\n")[0]  # from the previous cell
EXP_DIR = "/content/drive/MyDrive/Thesis/Experiments1"
NEW_YAML_F1 = f"{EXP_DIR}/yolo26n_exp2_F1weighted.yaml"

shutil.copy(STOCK_YAML, NEW_YAML_F1)

with open(NEW_YAML_F1) as f:
    arch = yaml.safe_load(f)

# head index 1 = absolute layer 12
target = arch['head'][1]
assert target == [[-1, 6], 1, 'Concat', [1]], (
    f"Unexpected layer at head[1]: {target} — this guess may not match your "
    f"actual stock yaml. Check the printed backbone/head from Cell 4's "
    f"output for the real layer 12 definition, then fix the assert and the "
    f"replacement line below before proceeding."
)

arch['head'][1] = [[-1, 6], 1, 'WeightedConcat', [1]]

with open(NEW_YAML_F1, 'w') as f:
    yaml.dump(arch, f, sort_keys=False, default_flow_style=None)

print("Modified head:")
for i, layer in enumerate(arch['head']):
    print(f"layer {11+i}: {layer}")
```
